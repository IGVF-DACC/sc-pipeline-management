```python
import firecloud.api as fapi
import pandas as pd
```

```python
with open("/Users/zheweishen/IGVF/IGVF_Repos/sc-pipeline-management/terra_datatables/all_submissions_under_terra.txt") as f:
    all_submission_ids = [entry.strip().split('/')[-2] for entry in f.readlines()]
```

```python
all_submission_ids
```

```python
fapi._set_session()
```

```python
submission_names = {}
for sub_id in all_submission_ids:
    curr_query = fapi.get_submission(workspace='IGVF Single-Cell Data Processing',
                                            namespace='DACC_ANVIL',
                                            submission_id=sub_id).json()
    curr_run_name = curr_query.get('submissionEntity', {}).get('entityName')
    curr_comment = curr_query.get('userComment')
    curr_config = curr_query.get('methodConfigurationName')
    curr_subdate = curr_query.get('submissionDate')
    curr_submitter = curr_query.get('submitter')
    submission_names[sub_id] = [curr_run_name, curr_comment, curr_config, curr_submitter, curr_subdate]
```

```python
final_table = pd.DataFrame(submission_names).T
final_table.columns = ['EntityName', 'User Comment', 'Pipeline', 'Submitters', 'Submitted On']
final_table.index.name = 'Submission IDs'
```
